```python
dbutils.widgets.text("table", "", "Source Table Name")
dbutils.widgets.text("schema", "", "Source Schema Name")
dbutils.widgets.text("catalog", "", "Source Catalog Name")
dbutils.widgets.text("destination_abfss", "", "Destination ABFSS Path (abfss://...)")
```

```python
sourceTable = dbutils.widgets.get("table")
sourceSchema = dbutils.widgets.get("schema")
sourceCatalog = dbutils.widgets.get("catalog")
destinationAbfss = dbutils.widgets.get("destination_abfss")
```

```python
data_export_df = spark.table(f"{sourceCatalog}.{sourceSchema}.{sourceTable}")
```

```python
now = datetime.now(timezone.utc).strftime('%Y%m%dT%H%MZ')
```

```python
data_export_df.write.format("parquet").save(destinationAbfss + f"/{sourceCatalog}/{sourceSchema}/{sourceTable}/{now}")
```
